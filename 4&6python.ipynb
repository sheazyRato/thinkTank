# Task 4

-Total Revenue by Genre (can be ordered by decade too)<br>
-Movie Count by Genre per Decade<br>
-Highest Earning Genre by Decade<br>
-Highest Rated Genre by Decade<br>
-Top Earning Genre for each Country (can be compared against top rated genre by country)

# Task 4 (Single Code Cell)
Takes 30-40s to execute on my laptop

```python
# setup
task4 = df.copy()

task4['genres'] = task4['genres'].str.split(', ')
task4 = task4.explode('genres')

task4 = task4[task4['genres'].notna()]#remove empty records




print("\n*******Total Revenue for each Genre*******")
genres = task4.groupby(['genres']).agg({
    'revenue': 'sum',
    'vote_average': 'mean',
    'runtime': 'mean'
})

genres.columns = ['total_rev', 'avg_rating', 'avg_runtime']
genres = genres.reset_index()
genres = genres.sort_values(['total_rev'], ascending=False)

print(genres)




print("\n*******Highest to Lowest Movie Count by Genre per Decade*******")
# by decade and genre with averages calculated 
stats = task4.groupby(['decade', 'genres']).agg({
    'id': 'count',           # number of movies
    # couldn't get average revenue to work
    'vote_average': 'mean'    # average rating
})

#renaming
stats.columns = ['movie_count', 'avg_rating']
stats = stats.reset_index()

stats = stats.sort_values(['decade', 'movie_count'], ascending=[True, False])

print(stats)




print("\n*******Highest Earning Genre by Decade*******")
# Group by decade and genre, calculate average revenue
earnings = task4.groupby(['decade', 'genres'])['revenue'].mean().round(2).reset_index()

# Sort by decade and then by revenue (highest to lowest)
earnings = earnings.sort_values(['decade', 'revenue'], ascending=[True, False])


for decade in sorted(earnings['decade'].unique()):
    top_genre = earnings[earnings['decade'] == decade].iloc[0]
    print(f"{int(decade)}s: {top_genre['genres']} - ${top_genre['revenue']:,.0f} average")




print("\n*******Highest Rated Genre by Decade*******")
# averaging the ratings
ratings = task4.groupby(['decade', 'genres'])['vote_average'].mean().round(2).reset_index()

# decade and then by rating (highest to lowest)
ratings = ratings.sort_values(['decade', 'vote_average'], ascending=[True, False])


for decade in sorted(ratings['decade'].unique()):
    top_genre = ratings[ratings['decade'] == decade].iloc[0]
    print(f"{int(decade)}s: {top_genre['genres']} - {top_genre['vote_average']} average rating")




print("\n*******Top Earning Genre for each Country*******")
countries = task4.copy()
countries['production_countries'] = countries['production_countries'].str.split(', ')
countries = countries.explode('production_countries')
countries = countries[countries['production_countries'].notna()]

countryGenre = countries.groupby(['production_countries', 'genres'])['revenue'].sum().reset_index()

countryGenre = countryGenre.sort_values(['production_countries', 'revenue'], ascending=[True, False])

for country in sorted(countryGenre['production_countries'].unique()):
    top_genre = countryGenre[countryGenre['production_countries'] == country].iloc[0]

    if top_genre['revenue'] > 0:
        
        amount = top_genre['revenue'] 
        print(f"{country}: {top_genre['genres']} - ${amount}")
```

```python
#setup
task6 = df.copy()


print("\n*******Genres by Country and Release Year*******")
task6['year'] = pd.to_datetime(task6['release_date']).dt.year

genreCountryYear = task6.groupby(['genres', 'production_countries', 'year']).agg({
    'id': 'count',
    'vote_average': 'mean',
    'revenue': 'mean'
}).round(2)

genreCountryYear.columns = ['movie_count', 'avg_rating', 'avg_revenue']
genreCountryYear = genreCountryYear.reset_index()

genreCountryYear = genreCountryYear.sort_values(['year', 'production_countries', 'genres'])

print(genreCountryYear.head(7))
print(f"\nTotal groups: {len(genreCountryYear)}")




print("\n*******Genres by Decade with Budget Bracket*******")
task6['budget_bracket'] = pd.cut(
    task6['budget'],
    bins=[0, 1000000, 10000000, 50000000, 100000000, float('inf')],
    labels=['<1M', '1M-10M', '10M-50M', '50M-100M', '>100M']
)

genrDecadeBudget = task6.groupby(['genres', 'decade', 'budget_bracket']).agg({
    'id': 'count',
    'vote_average': 'mean',
    'revenue': 'sum'
})

genrDecadeBudget.columns = ['movie_count', 'avg_rating', 'total_rev']
genrDecadeBudget = genrDecadeBudget.reset_index()

genrDecadeBudget = genrDecadeBudget[genrDecadeBudget['total_rev'] > 0]
genrDecadeBudget = genrDecadeBudget.sort_values(['decade', 'genres', 'budget_bracket'])

print(genrDecadeBudget.head(7))




print("\n*******Genre Performance for each Production Company*******")
companies = df.copy()

companies = companies.rename(columns = {'production_companies_clean': 'production_company'})
companies['production_company'] = companies['production_company'].str.strip('[]').str.replace("'", "").str.split(', ')
companies = companies.explode('production_company')

companyGenre = companies.copy()
companyGenre['genres'] = companyGenre['genres'].str.split(', ')
companyGenre = companies.explode('genres')

companyGenrePerf = companyGenre.groupby(
    ['production_company', 'genres']
).agg({
    'id': 'count',
    'vote_average': 'mean',
    'revenue': 'sum'
}).round(2)

companyGenrePerf.columns = ['movie_count', 'avg_rating', 'total_rev']
companyGenrePerf = companyGenrePerf.reset_index()

companyGenrePerf = companyGenrePerf[companyGenrePerf['total_rev'] > 0]

companyGenrePerf = companyGenrePerf.sort_values(['production_company', 'movie_count'],ascending=[True, False])

print(companyGenrePerf.head())




print("\n*******Multiple vs Single Genre Perfomance*******")

task6['num_genres'] = task6['genres'].str.split(', ').str.len()

genreCountStats = task6.groupby('num_genres').agg({
    'id': 'count',
    'revenue': 'mean',
    'vote_average': 'mean'
}).round(2)

genreCountStats.columns = ['movie_count', 'avg_rev', 'avg_rating']
genreCountStats = genreCountStats.reset_index()
genreCountStats = genreCountStats.sort_values('num_genres')

print(genreCountStats)




print("\n*******Movie Ratings by Spoken Lanuage in each Decade*******")
# I did not clean - 'No language'/'Unknown'
# some languages are not spelt in english 
langStats = task6.groupby(["decade", "main_spoken_language_clean"]).agg({
    'id': 'count',
    'vote_average': 'mean'
})

langStats.columns = ['movie_count', 'avg_rating']
langStats = langStats.reset_index()

langStats = langStats.sort_values(
    ["decade", "main_spoken_language_clean"]
)

print(langStats.head(20))




print("\n*******Highest Rated Genre by Country*******")
countryGenre = countries.groupby(['production_countries','genres'])['vote_average'].mean().round(2).reset_index()
countryGenre = countryGenre.sort_values(['production_countries', 'vote_average'], ascending = [True, False])

for country in sorted(countryGenre['production_countries'].unique()):
    top_rated = countryGenre[countryGenre['production_countries'] == country].iloc[0]
    print(f"{country}: {top_rated['genres']} - {top_rated['vote_average']} average rating")
```
